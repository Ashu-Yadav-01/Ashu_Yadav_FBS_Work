## linplot
1. trends over time
2. time series analysis
3. upward/downward trend

```python
import numpy as np
```

```python
overs = np.arange(1,21)
team_1 = np.random.randint(0,37,20)
team_2 = np.random.randint(0,37,20)
```

```python
team_1
```

```python
team_2
```

```python
team1_score = np.cumsum(team_1)
team1_score
```

```python
team2_score = np.cumsum(team_2)
```

```python
plt.figure(figsize=(7,3))
plt.style.use('dark_background')
plt.plot(overs , team1_score , marker='o', color='red', label='team_1')
plt.plot(overs , team2_score , marker='<', color='yellow', label='team_2')
plt.xlabel('overs')
plt.ylabel('score')
plt.title('match summary')
plt.legend()
plt.show()
```
