## bar plot
1. comparing categories
2. showing frequncies or count

```python
import matplotlib.pyplot as plt
```

```python
products=['hot coffee','cold coffe','tea','salad']
summer=[30,90,80,50]
```

```python
plt.figure(figsize=(7,3)) # width heigth
plt.bar(products,summer,color='black')
plt.xlabel('products')
plt.ylabel('count')
plt.show()
```
